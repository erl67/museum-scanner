```python
import os
import pandas as pd
import matplotlib.pyplot as plt
#os.getcwd()

df = pd.read_csv("file_counts_by_date.csv", parse_dates=['Date'])
df['Date_Label'] = pd.to_datetime(df['Date']).dt.strftime('%b %d').str.upper()

ax = df.plot(
    x='Date_Label', 
    y=['Scanned','Renamed'], 
    kind='bar', figsize=(12,6),
    color=['tab:blue', 'tab:orange']
)

ax.set_ylabel('Count')
ax.set_xlabel('')
ax.legend(title='')  # optional: cleaner legend
plt.tight_layout()
# plt.savefig('progress_split.png', dpi=200)
plt.show()

# Totals for legend
tot_scanned = df['Scanned'].sum()
tot_renamed = df['Renamed'].sum()

ax = df.plot(
    x='Date_Label',
    y=['Scanned', 'Renamed'],
    kind='bar',
    stacked=True,
    figsize=(12, 6),
    color=['tab:blue', 'tab:orange']
)

ax.set_ylabel('Count')
ax.set_xlabel('')

# Replace legend text with totals
handles, labels = ax.get_legend_handles_labels()
new_labels = [
    f"Scanned ({tot_scanned:,})",
    f"Renamed ({tot_renamed:,})"
]
ax.legend(handles, new_labels, title='')

plt.tight_layout()
plt.savefig('Progress.png', dpi=200)
plt.show()
```
